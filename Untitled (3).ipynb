```python
# ============================================================
# LAYER 2 (real): AraBERT fine-tuned hallucination classifier
# Trains on AraHalluEval's labeled answers. Runs on GPU (Lambda).
# ============================================================
!pip install -q transformers datasets accelerate scikit-learn

import csv, numpy as np, torch
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, roc_auc_score, f1_score, accuracy_score
from transformers import (AutoTokenizer, AutoModelForSequenceClassification,
                          TrainingArguments, Trainer)
from datasets import Dataset

MODEL = "aubmindlab/bert-base-arabertv2"
DATA  = "judge_train.csv"     # adjust path if needed (e.g. /content/... on Colab)

# --- label helpers (same as Layer 1) ---
LABEL_COLS = ['Named-Entity Hallucination','Temporal/Number Hallucination','Factual Contradiction',
              'Conflict Hallucination','K0wledge Source Conflict','Grammar Hallucination',
              'Generic/Imprecise Hallucination','Instruction Inconsistency','Code-Switching']
def true_label(r): return 1 if any(r.get(c,'').strip()=='1' for c in LABEL_COLS) else 0
def is_valid(r):   return all(r.get(c,'').strip() in ('0','1','') for c in LABEL_COLS)

# --- load labeled data ---
rows = [r for r in csv.DictReader(open(DATA, encoding="utf-8"))
        if r.get("answer","").strip() and is_valid(r)]
answers = [r["answer"] for r in rows]
golds   = [r["gold_answer"] for r in rows]
labels  = [true_label(r) for r in rows]
print(f"rows: {len(rows)} | positive rate: {np.mean(labels):.3f}")

# --- shared stratified split (seed 42, matches TF-IDF baseline) ---
idx = np.arange(len(rows))
tr, te = train_test_split(idx, test_size=0.2, random_state=42, stratify=labels)

tok = AutoTokenizer.from_pretrained(MODEL)

def run_variant(name, use_gold):
    print(f"\n{'='*60}\nVARIANT: {name}\n{'='*60}")

    def tokenize(batch):
        if use_gold:
            return tok(batch["gold"], batch["text"], truncation=True,
                       padding="max_length", max_length=512)
        return tok(batch["text"], truncation=True,
                   padding="max_length", max_length=512)

    train_ds = Dataset.from_dict({
        "text":  [answers[i] for i in tr],
        "gold":  [golds[i]   for i in tr],
        "label": [labels[i]  for i in tr],
    }).map(tokenize, batched=True)
    test_ds = Dataset.from_dict({
        "text":  [answers[i] for i in te],
        "gold":  [golds[i]   for i in te],
        "label": [labels[i]  for i in te],
    }).map(tokenize, batched=True)

    model = AutoModelForSequenceClassification.from_pretrained(MODEL, num_labels=2)

    args = TrainingArguments(
        output_dir=f"./arabert_judge_{name}",
        num_train_epochs=3,
        per_device_train_batch_size=16,
        per_device_eval_batch_size=32,
        eval_strategy="epoch",
        save_strategy="epoch",
        learning_rate=2e-5,
        weight_decay=0.01,
        logging_steps=20,
        load_best_model_at_end=True,
        metric_for_best_model="f1",
        report_to="none",
    )

    def metrics(p):
        preds = np.argmax(p.predictions, axis=1)
        return {"accuracy": accuracy_score(p.label_ids, preds),
                "f1": f1_score(p.label_ids, preds)}

    trainer = Trainer(model=model, args=args,
                      train_dataset=train_ds, eval_dataset=test_ds,
                      compute_metrics=metrics)
    trainer.train()

    out = trainer.predict(test_ds)
    probs = torch.softmax(torch.tensor(out.predictions), dim=1)[:,1].numpy()
    preds = np.argmax(out.predictions, axis=1)
    yte   = [labels[i] for i in te]

    print(f"\n--- {name} held-out ---")
    print(classification_report(yte, preds, target_names=["clean","hallucinated"], digits=3))
    print(f"ROC-AUC: {roc_auc_score(yte, probs):.3f}")

    trainer.save_model(f"./arabert_judge_{name}")
    tok.save_pretrained(f"./arabert_judge_{name}")
    return roc_auc_score(yte, probs)

auc_a = run_variant("answer_only", use_gold=False)
auc_b = run_variant("gold_answer", use_gold=True)

print(f"\n{'='*60}")
print(f"answer-only AUC : {auc_a:.3f}")
print(f"gold+answer AUC : {auc_b:.3f}")
print(f"baseline (TF-IDF): 0.82")
print("If gold+answer wins clearly, that's your Layer 2.")
```
